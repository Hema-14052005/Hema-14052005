```python
import matplotlib.pyplot as plt

expenses = []

# -------------------------
# Add Expense
# -------------------------
def add_expense():
    date = input("Enter date (DD-MM-YYYY): ")
    category = input("Enter category (Food, Travel, etc): ")
    description = input("Enter description: ")
    amount = float(input("Enter amount: "))

    expenses.append({
        "Date": date,
        "Category": category,
        "Description": description,
        "Amount": amount
    })

    print("Expense added successfully!\n")


# -------------------------
# View Expenses
# -------------------------
def view_expenses():
    if not expenses:
        print("No expenses recorded.\n")
        return

    print("\n--- All Expenses ---")
    for i, exp in enumerate(expenses, 1):
        print(f"{i}. {exp['Date']} | {exp['Category']} | {exp['Description']} | ₹{exp['Amount']}")
    print()


# -------------------------
# Total Expense
# -------------------------
def total_expense():
    total = sum(exp["Amount"] for exp in expenses)
    print(f"\nTotal Expense: ₹{total}\n")


# -------------------------
# Chart: Category-wise Bar Chart
# -------------------------
def category_chart():
    if not expenses:
        print("No expenses to show chart.\n")
        return

    category_totals = {}

    for exp in expenses:
        category = exp["Category"]
        category_totals[category] = category_totals.get(category, 0) + exp["Amount"]

    categories = list(category_totals.keys())
    amounts = list(category_totals.values())

    plt.figure()
    plt.bar(categories, amounts)
    plt.xlabel("Category")
    plt.ylabel("Amount")
    plt.title("Category-wise Expenses")
    plt.show()


# -------------------------
# Chart: Pie Chart
# -------------------------
def pie_chart():
    if not expenses:
        print("No expenses to show chart.\n")
        return

    labels = []
    sizes = []

    category_totals = {}
    for exp in expenses:
        category = exp["Category"]
        category_totals[category] = category_totals.get(category, 0) + exp["Amount"]

    for cat, amt in category_totals.items():
        labels.append(cat)
        sizes.append(amt)

    plt.figure()
    plt.pie(sizes, labels=labels, autopct="%1.1f%%")
    plt.title("Expense Distribution")
    plt.show()


# -------------------------
# Main Menu
# -------------------------
def main():
    while True:
        print("=== Expense Tracker ===")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. View Total Expense")
        print("4. Category-wise Bar Chart")
        print("5. Expense Pie Chart")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_expense()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            total_expense()
        elif choice == "4":
            category_chart()
        elif choice == "5":
            pie_chart()
        elif choice == "6":
            print("Goodbye!")
            break
        else:
            print("Invalid choice.\n")


# Run program
main()
```
